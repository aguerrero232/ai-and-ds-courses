## Population variance known z-score (3.9)


```python
import pandas as pd
import re
```

```python
data = pd.read_excel('3.9.Population-variance-known-z-score-lesson.xlsx', sheet_name='CI', skiprows=4, usecols=[0], names=['Price'])
```

```python
formatted = data['Price'].apply(lambda i: int(re.sub(r'[$,\\s]+', '', str(i))))
```

```python
data["Price"] = formatted
data
```

```python
sample_mean = data["Price"].mean()
sample_mean
```

```python
population_std = data["Price"].std()
population_std
```

```python

std_error = population_std / (len(data["Price"]) ** 0.5)
std_error
```

```python
ci_95_pct_lower = 1.96 * std_error
ci_95_pct_upper = 2.58 * std_error

print(ci_95_pct_lower, ci_95_pct_upper)
```

To calculate the 95% confidence interval you now just need to add/subtract the lower bounds to the sample mean.
To calculate the 99% confidence interval you now just need to add/subtract the upper bounds to the sample mean.

```python
ci_95_lower_bound = (sample_mean - ci_95_pct_lower).round(2)
ci_95_upper_bound = (sample_mean + ci_95_pct_lower).round(2)

ci_99_lower_bound = (sample_mean - ci_95_pct_upper).round(2)
ci_99_upper_bound = (sample_mean + ci_95_pct_upper).round(2)

print("95% confidence interval:", ci_95_lower_bound, ci_95_upper_bound)
print("99% confidence interval:", ci_99_lower_bound, ci_99_upper_bound)
```

## Central Limit Theroem

States that the no matter the distribution of the population the sampling distribution of the mean will be normally distributed.


sampling distribution ~ $N\left(\mu, \frac{\sigma^2}{n}\right)$, n > 30


## Standard Error

The **Standard Error** is a statistical measure that quantifies the amount of variability or dispersion in the sampling distribution of a statistic—most commonly, the **sample mean**.

$\text{Standard Error (SE)} = \frac{\sigma}{\sqrt{n}}$

Where:
- $\sigma$ = population standard deviation  
- n = sample size

## Estimators and Estimates

An approximation depending solely on sample information.
A specific value is an estimate. (confidence interval are preferred over estimates)

Point of Estimators and Estimates
Estimator | Parameter |  Estimate
----------|-----------|-------------
$\bar{x}$ | $\mu$ | Estimate Value
$s^2$ | $\sigma^2$ | Estimate Value

They both contain two properties:
* Bias: Has an expected value equal to the population parameter
* Efficiency: The difference between the estimator and the true value
