## Population variance known z-score (3.9)


```python
import pandas as pd
import re
```

```python
# read in xlsx starting at line 6
data = pd.read_excel('3.9.Population-variance-known-z-score-lesson.xlsx', sheet_name='CI', skiprows=4, usecols=[0], names=['Price'])
```

```python
formatted = data['Price'].apply(lambda i: int(re.sub(r'[$,\\s]+', '', str(i))))
```

```python
data["Price"] = formatted
data
```

```python
sample_mean = data["Price"].mean()
sample_mean
```

```python
population_std = data["Price"].std()
population_std
```

```python

std_error = population_std / (len(data["Price"]) ** 0.5)
std_error
```

```python
ci_95_pct_lower = 1.96 * std_error
ci_95_pct_upper = 2.58 * std_error

print(ci_95_pct_lower, ci_95_pct_upper)
```

To calculate the 95% confidence interval you now just need to add/subtract the lower bounds to the sample mean.
To calculate the 99% confidence interval you now just need to add/subtract the upper bounds to the sample mean.

```python
ci_95_lower_bound = sample_mean - ci_95_pct_lower
ci_95_upper_bound = sample_mean + ci_95_pct_lower

ci_99_lower_bound = sample_mean - ci_95_pct_upper
ci_99_upper_bound = sample_mean + ci_95_pct_upper

print(ci_95_lower_bound, ci_95_upper_bound)
print(ci_99_lower_bound, ci_99_upper_bound)
```
